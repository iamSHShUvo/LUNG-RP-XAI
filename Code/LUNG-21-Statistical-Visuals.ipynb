```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
from scipy import stats
import os

# --- CONFIGURATION ---
data_file = '../Results/final_radiomics_merged.csv' 
stats_file = '../Results/table_statistical_analysis.csv'
output_dir = '../Results/Plots_Statistics/'

os.makedirs(output_dir, exist_ok=True)

print("--- GENERATING MODERN STATISTICAL PLOTS (VIOLIN) ---")

# 1. Load Data
if not os.path.exists(data_file) or not os.path.exists(stats_file):
    print("Error: Input files missing.")
else:
    df = pd.read_csv(data_file)
    df_stats = pd.read_csv(stats_file)
    
    # Get Top 4 Most Significant Features
    top_features = df_stats.sort_values(by='P_Value').head(4)['Feature'].tolist()
    
    print(f"Visualizing: {top_features}")
    
    # 2. Generate Violin Plots
    for feature in top_features:
        clean_name = feature.replace('original_', '').replace('_', ' ')
        p_val = df_stats[df_stats['Feature'] == feature]['P_Value'].values[0]
        p_text = f"p < 0.001" if p_val < 0.001 else f"p = {p_val:.4f}"
        
        # Create Violin Plot
        fig = px.violin(
            df, 
            x='Cohort', 
            y=feature, 
            color='Cohort',
            box=True,       # Add internal boxplot
            points='all',   # Add jittered points
            hover_data=['PatientID'],
            title=f"<b>{clean_name}</b><br><sub>Significance: {p_text}</sub>",
            color_discrete_map={
                'Public (Western)': '#3498db', 
                'Local (Square Hospital)': '#e74c3c'
            }
        )
        
        # Modern Polish
        fig.update_layout(
            template='plotly_white',
            showlegend=False,
            width=700,
            height=600,
            yaxis_title=dict(text='Feature Value', font=dict(size=14, weight='bold')),
            xaxis_title='',
            title_x=0.5,
            font=dict(family="Arial", size=14, color="#333")
        )
        
        # Save HTML
        filename = f"{output_dir}Violin_{feature}.html"
        fig.write_html(filename)
        
        # Save PNG
        try:
            png_name = f"{output_dir}Violin_{feature}.png"
            fig.write_image(png_name, scale=3)
            print(f"Saved: {png_name}")
        except:
            pass
            
    print(f"\nAll plots saved to {output_dir}")
```

```python
import pandas as pd
import numpy as np
from scipy import stats
import plotly.express as px
import plotly.graph_objects as go
import os

# --- CONFIGURATION ---
input_file = '../Results/final_radiomics_merged.csv' 
output_csv = '../Results/table_statistical_analysis.csv'
output_dir = '../Results/Final_Figures/'
os.makedirs(output_dir, exist_ok=True)

print("--- AUTOMATED STATISTICAL AUDIT & VISUALIZATION ---")

# 1. Load Data
if not os.path.exists(input_file):
    print(f"ERROR: File not found: {input_file}")
else:
    df = pd.read_csv(input_file)
    print(f"Data Loaded: {len(df)} patients")
    
    # 2. Define Cohorts
    cohorts = df['Cohort'].unique()
    if len(cohorts) < 2:
        print("Error: Need at least 2 cohorts to compare.")
    else:
        group1_name = cohorts[0] # e.g. Public
        group2_name = cohorts[1] # e.g. Local
        print(f"\nComparing: '{group1_name}' vs. '{group2_name}'")

        group1 = df[df['Cohort'] == group1_name]
        group2 = df[df['Cohort'] == group2_name]

        features = [c for c in df.columns if 'original_' in c]
        print(f"Analyzing {len(features)} radiomic features...")

        results = []

        # 3. Statistical Loop
        for feature in features:
            v1 = group1[feature].dropna()
            v2 = group2[feature].dropna()
            
            # Descriptive
            mean1, sd1 = v1.mean(), v1.std()
            mean2, sd2 = v2.mean(), v2.std()
            
            # Normality & Significance
            try:
                if len(v1) > 3 and len(v2) > 3:
                    _, p_norm1 = stats.shapiro(v1)
                    _, p_norm2 = stats.shapiro(v2)
                    is_normal = (p_norm1 > 0.05) and (p_norm2 > 0.05)
                else:
                    is_normal = False
            except: is_normal = False 

            if is_normal:
                statistic, p_value = stats.ttest_ind(v1, v2, equal_var=False)
                test_used = "T-Test (Welch)"
            else:
                statistic, p_value = stats.mannwhitneyu(v1, v2)
                test_used = "Mann-Whitney U"
                
            # Fold Change (Log2 for Volcano Plot)
            # Avoid division by zero
            fold_change = mean2 / (mean1 + 1e-9)
            log2_fc = np.log2(fold_change + 1e-9)

            results.append({
                'Feature': feature,
                'Clean_Name': feature.replace('original_', '').replace('glcm_', 'Texture: ').replace('shape_', 'Shape: '),
                f'Mean ({group1_name})': f"{mean1:.2f} ± {sd1:.2f}",
                f'Mean ({group2_name})': f"{mean2:.2f} ± {sd2:.2f}",
                'Fold Change': fold_change,
                'Log2_FC': log2_fc,
                'P_Value': p_value,
                'Neg_Log10_P': -np.log10(p_value + 1e-300), # For plotting height
                'Test Used': test_used,
                'Significant': 'YES' if p_value < 0.05 else 'NO'
            })

        # 4. Save Table
        df_stats = pd.DataFrame(results)
        df_stats = df_stats.sort_values(by='P_Value')
        df_stats.to_csv(output_csv, index=False)
        print(f"Stats saved to: {output_csv}")

        # --- 5. VISUALIZATION: THE VOLCANO PLOT ---
        print("\nGenerating Volcano Plot...")
        
        fig_volcano = px.scatter(
            df_stats, 
            x='Log2_FC', 
            y='Neg_Log10_P',
            color='Significant',
            hover_data=['Clean_Name', 'P_Value'],
            color_discrete_map={'YES': '#e74c3c', 'NO': 'grey'},
            title=f"<b>Volcano Plot: Domain Shift Magnitude</b><br><sup>{group1_name} vs. {group2_name}</sup>",
            labels={'Log2_FC': 'Log2 Fold Change (Effect Size)', 'Neg_Log10_P': '-Log10 P-Value (Significance)'}
        )
        
        # Add threshold lines
        fig_volcano.add_hline(y=-np.log10(0.05), line_dash="dash", line_color="black", annotation_text="p=0.05")
        fig_volcano.add_vline(x=1, line_dash="dot", line_color="grey")
        fig_volcano.add_vline(x=-1, line_dash="dot", line_color="grey")
        
        fig_volcano.update_layout(template='plotly_white', width=800, height=600)
        fig_volcano.write_html(f"{output_dir}Figure_Volcano_Plot.html")
        
        try:
            fig_volcano.write_image(f"{output_dir}Figure_Volcano_Plot.png", scale=3)
            print("  -> Volcano Plot saved.")
        except: pass
        fig_volcano.show()

        # --- 6. VISUALIZATION: HEATMAP OF TOP FEATURES ---
        print("\nGenerating Top Features Heatmap...")
        
        # Get top 10 features
        top_features = df_stats.head(10)['Feature'].tolist()
        clean_top_names = df_stats.head(10)['Clean_Name'].tolist()
        
        # Filter original data for heatmap
        # Normalize data (Z-score) for better visualization
        df_heatmap = df[['Cohort'] + top_features].copy()
        
        # We need to map feature names to clean names
        df_heatmap.columns = ['Cohort'] + clean_top_names
        
        # Calculate mean value per cohort for the heatmap (aggregated view)
        heatmap_data = df_heatmap.groupby('Cohort').mean().T
        
        fig_heat = px.imshow(
            heatmap_data,
            color_continuous_scale='RdBu_r', # Red-Blue Diverging
            aspect="auto",
            title="<b>Heatmap: Top 10 Divergent Features</b><br><sup>Normalized Mean Values</sup>",
            labels=dict(x="Cohort", y="Radiomic Feature", color="Z-Score Mean")
        )
        
        fig_heat.update_layout(template='plotly_white', width=600, height=600)
        fig_heat.write_html(f"{output_dir}Figure_Heatmap_TopFeatures.html")
        
        try:
            fig_heat.write_image(f"{output_dir}Figure_Heatmap_TopFeatures.png", scale=3)
            print("  -> Heatmap saved.")
        except: pass
        fig_heat.show()

        print("\nAll statistical visuals generated.")
```
