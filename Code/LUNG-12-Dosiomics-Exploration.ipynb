```python
import os
import pydicom


# --- Forensic DICOM Inspector ---


data_path = '../Data/NSCLC-Radiomics/'
patient_id = 'LUNG1-001'
patient_folder = os.path.join(data_path, patient_id)


print(f"--- Inspecting all files for patient: {patient_id} ---")
for root, _, files in os.walk(patient_folder):
    for file in files:
        file_path = os.path.join(root, file)
        try:
            # Read the file header to inspect its tags
            dcm = pydicom.dcmread(file_path, stop_before_pixels=True)


            # Safely get the Modality tag
            modality = getattr(dcm, 'Modality', 'N/A')


            print(f"File: {file:<10}  -->  Modality: {modality}")


        except Exception:
            # This will catch any non-DICOM files
            continue
```
